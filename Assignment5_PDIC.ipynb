```python
#1. Repeat a tuple three times using * operator

t = (1, 2, 3)

print(t * 3)
```

```python
#2. Join three tuples using + operator

t1 = (1, 2)
t2 = (3, 4)
t3 = (5, 6)

print(t1 + t2 + t3)
```

```python
#3. Check whether an element exists in tuple

t = (10, 20, 30, 40)

x = 30

if x in t:
    print("Element Found")
else:
    print("Element Not Found")
```

```python
#4. Find total, highest and lowest value without sum(), max(), min()

t = (10, 20, 30, 40, 50)

total = 0
highest = t[0]
lowest = t[0]

for i in t:
    total = total + i

    if i > highest:
        highest = i

    if i < lowest:
        lowest = i

print("Total =", total)
print("Highest =", highest)
print("Lowest =", lowest)
```

```python
#5. Keep only values greater than 10

t = (3, 14, 7, 22, 9, 41, 18, 5)

result = ()

for i in t:
    if i > 10:
        result = result + (i,)

print(result)
```

```python
#6. Count elements in set without len()

s = {"cat", "dog", "bird", "fish"}

count = 0

for i in s:
    count += 1

print("Total Elements =", count)
```

```python
#7. Combine two sets

s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}

print(s1.union(s2))
```

```python
#8. Find common elements in two sets

s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}

print(s1.intersection(s2))
```

```python
#9. Find elements that are in either set but not both

s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}

print(s1.symmetric_difference(s2))
```
