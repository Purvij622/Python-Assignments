```python
#1. Replace NaN with 0 and interchange rows and columns

import numpy as np

a = np.array([[6, -8, 73, -110],
              [np.nan, -8, 0, 94]])

a = np.nan_to_num(a, nan=0)

print("NaN Replaced:")
print(a)

print("Rows and Columns Interchanged:")
print(a.T)
```

```python
#2. Move axes of 3D array to new positions

import numpy as np

a = np.arange(24).reshape(2, 3, 4)

print("Original Shape =", a.shape)

b = np.moveaxis(a, 0, 2)

print("New Shape =", b.shape)
print(b)
```

```python
#3. Replace NaN values with average of columns

import numpy as np

a = np.array([[10, np.nan, 30],
              [20, 40, np.nan],
              [30, 50, 60]])

col_mean = np.nanmean(a, axis=0)

inds = np.where(np.isnan(a))

a[inds] = np.take(col_mean, inds[1])

print(a)
```

```python
#4. Replace negative values with zero

import numpy as np

a = np.array([10, -5, 20, -8, 15])

a = np.where(a < 0, 0, a)

print(a)
```

```python
#5. Calculate average, mean, median and mode

import numpy as np
from statistics import mode

arr1 = np.array([[3, 4],
                 [5, 6]])

arr2 = np.array([[1, 0],
                 [7, 8]])

avg = (arr1 + arr2) / 2

print("Average Array")
print(avg)

a = np.concatenate((arr1.flatten(),
                    arr2.flatten()))

print("Mean =", np.mean(a))
print("Median =", np.median(a))
print("Mode =", mode(a))
```

```python
#6. Solve equations using linalg() and inverse matrix

import numpy as np

A = np.array([
    [1, -2, 3],
    [-1, 3, -1],
    [2, -5, 5]
])

B = np.array([9, -6, 17])

X = np.linalg.solve(A, B)

print("Using linalg.solve()")
print(X)

A_inv = np.linalg.inv(A)

X2 = np.dot(A_inv, B)

print("Using Inverse Matrix")
print(X2)
```
